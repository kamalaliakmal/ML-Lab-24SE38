### Task 07: Create a function to extract the domain from "user@domain.com".

```python
def get_domain(email):
    return email.split('@')[1]

print("Q7: ",get_domain("user@domain.com"))
```

### Task 08: Count occurrences of 'dog' in a string (case-insensitive)

```python
def find_dog(text):
    return "dog" in text.lower()

print("Q8: ",find_dog("There is a DOG here"))
print(find_dog("I have a cat"))
```

### Task 09: Count how many times "dog" occurs in a string

```python
def count_dog(text):
    return text.lower().count("dog")

print("Q9: ",count_dog("dog Dog DOG cat dog"))
```

### Task 10: Implement the caught_speeding(speed, is_birthday): ≤ 60 "No Ticket", 61–80 "Small Ticket", ≥ 81 "Big Ticket". On a birthday, allow 5 mph more.

```python
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed -= 5

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"


print("Q10: ",caught_speeding(60, False))
print(caught_speeding(75, True))
print(caught_speeding(85, False))
print(caught_speeding(90, True))
```
